## Paso 1: Configurar el Entorno Conda

Sigue estos simples pasos para configurar tu entorno Conda:

1. **Crear el Entorno Conda**:
   - Abre tu terminal o símbolo del sistema.
   - Navega al directorio del repositorio:
     ```bash
     cd ruta/a/tu/repositorio
     ```
   - Crea el entorno usando el archivo `environment.yaml`:
     ```bash
     conda env create -f environment.yaml
     ```
   - Este comando configura el entorno Conda con todos los paquetes necesarios.

2. **Activar el Entorno**:
   - Activa el entorno:
     ```bash
     conda activate ai-engineering-bootcamp
     ```

Tu entorno de desarrollo está ahora listo con la versión de Python requerida y los paquetes.

> **Nota**: Si prefieres o si no estás usando Conda, puedes instalar los paquetes requeridos usando `pip`:
  ```bash
  pip install -r requirements.txt
  ```

## Paso 3: Configurar Servicios de Azure para Soluciones de IA

Para construir y desplegar efectivamente tus soluciones de IA, es esencial configurar varios servicios de Azure. Esta guía proporciona instrucciones paso a paso para crear cada servicio

#### 1. Azure AI Foundry

**Propósito**: Azure AI Foundry ofrece un entorno integrado para construir, probar y desplegar modelos y aplicaciones de IA.

**Pasos**:

- **Acceder a Azure AI Foundry**:
  - Navega al portal de [Azure AI Foundry](https://ai.azure.com/).
  - Inicia sesión con tus credenciales de Azure.

- **Crear un Nuevo Proyecto**:
  - Haz clic en **"Create a new project"**.
  - Proporciona un nombre de proyecto, selecciona la suscripción apropiada, y elige o crea un grupo de recursos.
  - Selecciona la región deseada y haz clic en **"Create"**.

Para orientación detallada, consulta la documentación oficial: [¿Qué es Azure AI Foundry?](https://learn.microsoft.com/en-us/azure/ai-studio/what-is-ai-studio#how-to-get-access)

#### 2. Azure AI Search

**Propósito**: Azure AI Search es un servicio de búsqueda en la nube que proporciona capacidades de búsqueda poderosas y sofisticadas para tus aplicaciones.

**Pasos**:

- **Crear un Servicio de Azure AI Search**:
  - Inicia sesión en el [portal de Azure](https://portal.azure.com/).
  - Haz clic en **"Create a resource"** y busca **"Azure AI Search"**.
  - Selecciona **"Azure AI Search"** de los resultados y haz clic en **"Create"**.

- **Configurar el Servicio**:
  - En la pestaña **Basics**, proporciona la siguiente información:
    - **Subscription**: Selecciona tu suscripción de Azure.
    - **Resource Group**: Elige un grupo de recursos existente o crea uno nuevo.
    - **Service Name**: Introduce un nombre único para tu servicio de búsqueda.
    - **Region**: Selecciona la región más cercana a tus usuarios.
    - **Pricing Tier**: Nota que el tier gratuito tiene limitaciones; considera el tier **Basic**. 

- **Revisar y Crear**:
  - Haz clic en **"Review + create"**, verifica los detalles, y luego haz clic en **"Create"** para desplegar el servicio.

Para una guía paso a paso, consulta la guía oficial: [Crear un servicio de búsqueda en el portal de Azure](https://learn.microsoft.com/en-us/azure/search/search-create-service-portal)

#### 3. Azure Blob Storage

**Propósito**: Azure Blob Storage proporciona almacenamiento de objetos escalable y costo-efectivo para datos no estructurados, como documentos, imágenes, videos, respaldos y logs.

**Pasos**:

- **Crear una Cuenta de Almacenamiento**:
  - Inicia sesión en el [portal de Azure](https://portal.azure.com/).
  - Haz clic en **"Create a resource"**, luego selecciona **"Storage account"**.
  - En la pestaña **Basics**, proporciona la siguiente información:
    - **Subscription**: Selecciona tu suscripción de Azure.
    - **Resource Group**: Elige un grupo de recursos existente o crea uno nuevo.
    - **Storage Account Name**: Introduce un nombre único para tu cuenta de almacenamiento.
    - **Region**: Selecciona la región más cercana a tus usuarios.
    - **Performance**: Elige **Standard** para almacenamiento costo-efectivo.
    - **Redundancy**: Selecciona **Locally-redundant storage (LRS)** para la opción de menor costo.
    - Ve a la pestaña **Advanced** y habilita **Allow Enabling Anonymous Access on Individual Containers**. Esta característica te permite habilitar acceso anónimo a contenedores específicos dentro de tu Azure Blob Storage (escenarios de POC y pruebas).

- **Revisar y Crear**:
  - Haz clic en **"Review + create"**, verifica los detalles, y luego haz clic en **"Create"** para desplegar la cuenta de almacenamiento.

- **Crear un Contenedor Blob**:
  - Después de que la cuenta de almacenamiento sea creada, navega a ella.
  - En el menú de la izquierda, bajo **Data storage**, selecciona **Containers**.
  - Haz clic en **"+ Container"**, proporciona un nombre, establece el **Public access level** según sea necesario, y haz clic en **"Create"**.

Para instrucciones detalladas, consulta: [Crear una cuenta de almacenamiento de Azure](https://learn.microsoft.com/en-us/azure/storage/common/storage-account-create)

#### 4. Azure AI Document Intelligence

**Propósito**: Azure AI Document Intelligence utiliza aprendizaje automático avanzado para extraer automáticamente y con precisión texto, pares clave-valor, tablas y estructuras de documentos.

## Paso 5: Configurar Variables de Entorno para todos los Servicios de Azure requeridos

Para asegurar integración perfecta con servicios de Azure, necesitas configurar variables de entorno en un archivo `.env`. Esta configuración incluirá credenciales esenciales para Azure OpenAI, Azure Search, Azure Blob Storage, Azure AI Services, y Azure Document Intelligence.

## Configuración de Variables de Entorno

### Pasos

1. **Crear el Archivo `.env`**:
   - En el directorio raíz de tu proyecto, crea un nuevo archivo llamado `.env`.

2. **Agregar las Siguientes Variables**:
   - Abre el archivo `.env` en un editor de texto y agrega las siguientes variables de entorno, reemplazando los valores de marcador de posición con tus credenciales reales de servicios de Azure:

     ```env
     # Configuración de Azure OpenAI
     AZURE_OPENAI_KEY=<TU CLAVE DE AZURE OPENAI>
     AZURE_OPENAI_ENDPOINT=<TU ENDPOINT DE AZURE OPENAI>
     AZURE_OPENAI_API_VERSION=<TU VERSIÓN API DE AZURE OPENAI>
     AZURE_OPENAI_EMBEDDING_DEPLOYMENT=<TU DESPLIEGUE DE EMBEDDING DE AZURE OPENAI>
     AZURE_OPENAI_CHAT_DEPLOYMENT_ID=<TU ID DE DESPLIEGUE DE CHAT DE AZURE OPENAI>
     AZURE_OPENAI_EMBEDDING_DIMENSIONS=<TUS DIMENSIONES DE EMBEDDING DE AZURE OPENAI>

     # Configuración de Azure AI Search
     AZURE_SEARCH_SERVICE_NAME=<TU NOMBRE DE SERVICIO DE AZURE SEARCH>
     AZURE_SEARCH_INDEX_NAME=<TU NOMBRE DE ÍNDICE DE AZURE SEARCH>
     AZURE_AI_SEARCH_ADMIN_KEY=<TU CLAVE ADMIN DE AZURE SEARCH>
     AZURE_AI_SEARCH_SERVICE_ENDPOINT=<TU ENDPOINT DE SERVICIO DE AZURE SEARCH>

     # Configuración de Azure Blob Storage
     AZURE_STORAGE_ACCOUNT_NAME=<TU NOMBRE DE CUENTA DE AZURE STORAGE>
     AZURE_STORAGE_ACCOUNT_KEY=<TU CLAVE DE CUENTA DE AZURE STORAGE>
     AZURE_STORAGE_CONNECTION_STRING=<TU CADENA DE CONEXIÓN DE AZURE STORAGE>
     AZURE_BLOB_CONTAINER_NAME=<TU NOMBRE DE CONTENEDOR DE AZURE BLOB>

     # Recurso Multi-Servicio de Azure AI Services
     AZURE_AI_SERVICES_KEY=<TU CLAVE DE AZURE AI SERVICES>

     # Configuración de API de Azure Document Intelligence
     AZURE_DOCUMENT_INTELLIGENCE_ENDPOINT=<TU ENDPOINT DE AZURE DOCUMENT INTELLIGENCE>
     AZURE_DOCUMENT_INTELLIGENCE_KEY=<TU CLAVE DE AZURE DOCUMENT INTELLIGENCE>
     ```

### Reemplazar Valores de Marcador de Posición

- `<TU CLAVE DE AZURE OPENAI>`: Clave API para tu servicio de Azure OpenAI.
- `<TU ENDPOINT DE AZURE OPENAI>`: URL del endpoint para tu servicio de Azure OpenAI.
- `<TU VERSIÓN API DE AZURE OPENAI>`: Versión API para Azure OpenAI (ej., `2023-06-01-preview`).
- `<TU DESPLIEGUE DE EMBEDDING DE AZURE OPENAI>`: Nombre de tu despliegue de embedding de Azure OpenAI.
- `<TU ID DE DESPLIEGUE DE CHAT DE AZURE OPENAI>`: ID de despliegue para el modelo de chat de Azure OpenAI.
- `<TUS DIMENSIONES DE EMBEDDING DE AZURE OPENAI>`: Dimensiones de embedding para tu despliegue.

- `<TU NOMBRE DE SERVICIO DE AZURE SEARCH>`: Nombre de tu servicio de Azure AI Search.
- `<TU NOMBRE DE ÍNDICE DE AZURE SEARCH>`: Nombre de tu índice de Azure AI Search.
- `<TU CLAVE ADMIN DE AZURE SEARCH>`: Clave admin para servicio de Azure AI Search.
- `<TU ENDPOINT DE SERVICIO DE AZURE SEARCH>`: URL del endpoint para tu servicio de Azure AI Search.

- `<TU NOMBRE DE CUENTA DE AZURE STORAGE>`: Nombre de tu cuenta de Azure Storage.
- `<TU CLAVE DE CUENTA DE AZURE STORAGE>`: Clave para tu cuenta de Azure Storage.
- `<TU CADENA DE CONEXIÓN DE AZURE STORAGE>`: Cadena de conexión para Azure Storage.
- `<TU NOMBRE DE CONTENEDOR DE AZURE BLOB>`: Nombre de tu contenedor de Azure Blob Storage.

- `<TU CLAVE DE AZURE AI SERVICES>`: Clave API para tu Recurso Multi-Servicio de Azure AI.

- `<TU ENDPOINT DE AZURE DOCUMENT INTELLIGENCE>`: URL del endpoint para Azure AI Document Intelligence.
- `<TU CLAVE DE AZURE DOCUMENT INTELLIGENCE>`: Clave API para Azure AI Document Intelligence.

#### Asegurar tu Archivo `.env`

- Asegúrate de que tu archivo `.env` esté listado en tu archivo `.gitignore` para prevenir que sea subido a sistemas de control de versiones.

````python
import os
import base64
import mimetypes
from openai import AzureOpenAI
from dotenv import load_dotenv

# Cargar variables de entorno del archivo .env
load_dotenv(dotenv_path="./.env")

# Recuperar variables de entorno
api_key = os.getenv("AZURE_OPENAI_KEY")
deployment_id = os.getenv("AZURE_OPENAI_CHAT_DEPLOYMENT_ID")
api_endpoint = os.getenv("AZURE_OPENAI_ENDPOINT")
api_version = os.getenv("AZURE_OPENAI_API_VERSION")

# Inicializar el cliente AzureOpenAI
client = AzureOpenAI(
    api_key=api_key,
    azure_endpoint=api_endpoint,
    api_version=api_version
)

# Ruta a la imagen
image_path = 'utils/images/azure_logo.png'

# Preparar el mensaje del usuario con contenido de texto e imagen
user_message = {
    "role": "user",
    "content": [{"type": "text", "text": "Describe la siguiente imagen:"}],
}

# Codificar la imagen como base64 y agregar al mensaje del usuario
try:
    with open(image_path, "rb") as image_file:
        encoded_image = base64.b64encode(image_file.read()).decode("utf-8")
        mime_type, _ = mimetypes.guess_type(image_path)
        mime_type = mime_type or "application/octet-stream"
        user_message["content"].append(
            {
                "type": "image_url",
                "image_url": {
                    "url": f"data:{mime_type};base64,{encoded_image}",
                },
            }
        )
except Exception as e:
    print(f"Error procesando imagen {image_path}: {e}")
    exit(1)

# Preparar el payload de mensajes
messages = [
    {
        "role": "system",
        "content": "Eres un asistente de IA que ayuda a describir imágenes usando tecnología OCR."
    },
    user_message
]

# Enviar la solicitud y manejar la respuesta
try:
    response = client.chat.completions.create(
        model=deployment_id,
        messages=messages,
        temperature=0.7,
        top_p=0.95,
        max_tokens=800
    )
    print("Respuesta de Azure OpenAI:")
    print(response.choices[0].message.content)
except Exception as e:
    print(f"Ocurrió un error: {e}")
````
